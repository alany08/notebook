if A is a transformation in the standard basis C, and then the basis B is given by a span of vectors span{b1...bn} then the change of basis from B to C (P) is \[ b1 ... bn \]. To turn the vector into our basis before applying the transformation in our basis, we need to multiply the changed vector by A, so instead of Ax we do APx. But then to finally express the transformation, or output vector in their basis, we need to change back, giving P^-1 A P as the final B-matrix transform.
