### Tensors
These behave sort of like matricies in linear algebra, mapping a set of vectors to another. The dimension of each vector isn't explicitly encoded. Each tensor has a upper index, and a lower index. The upper index is called the **contravariant**, while the lower is called the **covariant**. They represent the dimensionality that they operate in.
