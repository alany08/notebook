## Electric Dipole
At $z >> d$, a pair of charges ($-q$ and $+q$) becomes a center at which the effective force can be approximated to. The electric dipole moment can be measured by the separation ($d$) and the charge ($q$) of the dipole, and is represented by $\vec{p}$ which is calculated $qd$. This effectively shows the polarization of the dipole. The larger the separation $d$, at distances $z >> d$, the more effect this dipole has. The stronger the charge, the same effect.

In an electric field, there will be a torque exerted upon this dipole. The moment of the dipole $\vec{p}$ describes how polarized this dipole is, and points from the negative to positive charge. Let there be an electric field $\vec{E}$ that is acting upon this dipole. Then, there will be a torque exerted upon this dipole due to their opposing charges. Because the force upon the individual charges $q$ are parallel to the electric field, and the separation $d$ is acting as a lever arm around the center of mass, a torque $\vec{\tau}$ is exerted on the dipole system, which is usually $\vec{\tau}=\vec{r} \times \vec{F}$. Because the dipole moment is analogous to a lever arm and charge, and the electric field is analogous to force per charge, we can replace $\vec{r}$ and $\vec{F}$ with $\vec{p}$ and $\vec{E}$ giving
$$
\vec{\tau} = \vec{p} \times \vec{E}
$$

**Here it is a bit clearer:**

Basically, dipole moment $\vec{p}$ is proportional the the charge, and the distance between them (think moments for center of mass, like $mx$ or $my$). This we can find the torque to act on the lever arm of $\frac{\vec{d}}{2}$ in both the upper and lower charge in opposing directions, but both in the same in terms of angular motion. Because the force in the electric field $\vec{E}$ is equivalent to $\vec{F}=q\vec{E}$, the torque on one charge is equivalent to

$$
    \vec{\tau} = \vec{r} \times \vec{F}
$$

which, when plugging in force and the lever arm simplifies to

$$
    \vec{\tau} = (
        \frac{\vec{d}}{2}
    ) \times (
        q\vec{E}
    )
$$

multiply by two for torque from both charges, pull out the non vector values, we get

$$
    \vec{\tau} = \vec{d}q \times \vec{E}
$$

$$
    \vec{\tau} = \vec{p} \times \vec{E}
$$

## Charges Over a Continuous Shape
Instead of certain point charges, these calculations aim to calculate electric force on a point over a continuous surface or shape, such as a long thin wire of equally distributed charge. This starts to require calculus.

#### Ring
A distribution of charge over a ring of radius $R$, at a distance $z$ away from the point $P$. Because this is a ring, we will find it easier to work in polar coordinates around the ring.

First of all, recall Coulomb's law
$$
F = k \frac{q_1 q_2}{r^2}
$$
where $r$ is the distance between two charges.

In this problem, we will represent the charge of P with $q_P$ and the total charge of the ring as $q_r$. So let's set up our integrand. The charge at a differential of the ring, or an infinitely small part of the ring can be represented by $\frac{q_r}{2\pi}$, as we are splitting the charge into each little angle. The distance between the small chunk of ring and $P$, with a little bit of trigonometry, we can find to be $\sqrt{R^2+z^2}$, and this remains constant. Plugging the previous two terms we found, the electric force formula is
$$
F = k 
    \frac{
        \frac{q_r}{2\pi} q_P
    }
    {
        \sqrt{R^2+z^2}^2
    }
$$
$$
F = k 
    \frac{
        \frac{q_r}{2\pi} q_P
    }
    {
        R^2+z^2
    }
$$
Then we integrate that all over the whole circle, which is $2\pi$ radians.
$$
F = \int_{0}^{2\pi} 
    k \frac{
        \frac{q_r}{2\pi} q_P
    }
    {
        R^2+z^2
    }d\theta
$$
Evaluate this, and you notice that because theres no $\theta$ it's literally
$$
F = \frac{
        q_r q_P
    }
    {
        R^2+z^2
    }
$$
why did we do calculus? However there is something wrong with this. The component of the force perpendicular to $z$ cancels out, leaving the magnitude to be multiplied by $\frac{z}{\sqrt{R^2+z^2}}$ and then our final formula is
$$
F = \frac{
        q_r q_P
    }
    {
        (R^2+z^2)^{\frac{3}{2}}
    }
$$
For arcs that are not a full ring, you can change the bounds of the integral a few steps above and re-evaluate, but in general it should be angle of the arc/$2\pi$

# Electric Flux
Flux is defined as the amount of electric field that goes through a surface, and it's a dot product of the electric field with the surface $E \cdot A$. For a non-uniform surface, you do a surface integral. Here's Gauss's Law, stating total electric flux in a closed surface is equal to the charge enclosed divided by permittivity.
$$
\boxed{
    \Phi =
        \oint \vec{E} \cdot d\vec{A} =
        \frac{
            Q_{enc}
        }{
            \epsilon _0
        }
}
$$
with $\vec{A}$ as the normal vector for the surface, and $\vec{E}$ being the electric field. If a closed surface does not have any internal charge, then every field line that enters the surface also exits, therfore the net flux will be zero.

Recall that $\epsilon _0$ defines $k$, the charge constant as $k = \frac{1}{4\pi \epsilon _0}$, but why? Let's derive this with a charge $q$ enclosed by a sphere. It will create an electric field 
$$
    \vec{E} = k\frac{q}{r^2}
$$
where $r$ defines the radius of the sphere. Assuming the system is symetric, $\vec{E}$ will remain normal to the surface of the sphere $\vec{A}$. Let's do some work on the flux integral first.
$$
    \oint \vec{E} \cdot d\vec{A} = \frac{Q_{enc}}{\epsilon _0}
$$
because $\vec{E}$ is constant and normal to $\vec{A}$ this means we can pull it out of the integral
$$
    E \oint d\vec{A} = \frac{Q_{enc}}{\epsilon _0}
$$
integral of nothing is just $x$ or in this case $A$
$$
    EA = \frac{Q_{enc}}{\epsilon _0}
$$
$A$ represents area, and an area of a sphere is $4 \pi r^2$ so
$$
    4 \pi r^2 E = \frac{Q_{enc}}{\epsilon _0}
$$
$$
    E = \frac{Q_{enc}}{4 \pi r^2 \epsilon _0}
$$
$E$ is also defined above as $k\frac{q}{r^2}$ so let's set them equal
$$
    E = \frac{Q_{enc}}{4 \pi r^2 \epsilon _0} = k\frac{q}{r^2}
$$
(a little variable naming, $Q_{enc}$ is just $q$)
$$
    \frac{1}{4 \pi r^2 \epsilon _0} = k\frac{1}{r^2}
$$
$$
    k = \frac{1}{4 \pi \epsilon _0}
$$
and there we go, the derivation of the $k$ constant.
